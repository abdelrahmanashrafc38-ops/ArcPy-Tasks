## Place populations by State / Providence
What are the Populations of each State/Province in One Country?

#### First we select the country by using its name as stored in the ADMNAME field:

```python
arcpy.env.overwriteOutput = True
arcpy.env.workspace = r"D:\Learning\iti\CoursesDoc\GIS\Course10_ArcPY\LPA\Data\Sample.gdb"
placesFC = r"D:\Learning\iti\CoursesDoc\GIS\Course10_ArcPY\LPA\Data\ne_10m_populated_places.shp"
arcpy.Select_analysis(placesFC,"Places1Country","ADM0NAME = 'Uganda'")
```

#### Now we will summerize the population by state or province and sort the result Asc:

```python
arcpy.analysis.Statistics("Places1Country","Places1CountryStates",[["pop_Max","sum"]],"ADM1NAME")
arcpy.management.Sort("Places1CountryStates","Places1CountryStatesSorted",[["SUM_POP_MAX","Ascending"]])
```
